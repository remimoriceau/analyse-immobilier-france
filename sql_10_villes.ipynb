```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("immobilier_paris.db")
df = pd.read_csv("C:\econometrie2\projet S2\immobillier\immobilier_10villes_2024.csv")
df.to_sql("transactions", conn, if_exists="replace", index=False)

print("Base créée :", pd.read_sql("SELECT COUNT(*) as total FROM transactions", conn))
```

```python
requete1 = pd.read_sql("""
    SELECT 
        Ville,
        ROUND(AVG("Prix au m2"), 0) AS prix_moyen_m2,
        COUNT(*) AS nb_ventes
    FROM transactions
    GROUP BY Ville
    ORDER BY prix_moyen_m2 DESC
""", conn)

print(requete1)
```

```python
print(pd.read_sql("SELECT * FROM transactions LIMIT 1", conn).columns.tolist())
```

```python
requete2 = pd.read_sql("""
    SELECT 
        Ville,
        COUNT(*) AS nb_ventes
    FROM transactions
    GROUP BY Ville
    HAVING nb_ventes > 25000
    ORDER BY nb_ventes DESC
""", conn)

print(requete2)
```

```python
requete3 = pd.read_sql("""
    SELECT 
        "Type local",
        ROUND(AVG("Prix au m2"), 0) AS prix_moyen_m2,
        COUNT(*) AS nb_ventes
    FROM transactions
    GROUP BY "Type local"
""", conn)

print(requete3)
```

```python
requete4 = pd.read_sql("""
    SELECT 
        Ville,
        "Type local",
        "Valeur fonciere",
        "Surface reelle bati",
        ROUND("Prix au m2", 0) AS prix_m2
    FROM transactions
    ORDER BY "Valeur fonciere" DESC
    LIMIT 5
""", conn)

print(requete4)
```
